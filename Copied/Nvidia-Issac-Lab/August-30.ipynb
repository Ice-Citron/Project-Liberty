   The script is a small ROS 2 diagnostic code. It watches the TF tree during
   a simulation run. It writes one JSON line per tick at 10 Hz into one JSONL
   file. Each line contains a timestamp, the plug-tip pose in the world frame,
   the TCP pose in the world frame, the calculated TCP-to-plug-tip transform,
   and the z height of every cable link. The cable link frames appear in TF
   only when you run with `...` ... script checks for them at startup ... only 
   a development tool. MyCheatCode will never import it or read its output 
   while running, so it does not break any rules

   The script measures the two numbers required by the M1.3 design. The first
   is the hang depth: the distance that the lowest part of the cable hangs below
   the plug while the arm moves at height. This sets the safe traverse height 
   for the approach stage. The corridor must clear the card tops b at least
   the hang depth. Otherwise, the cable may catch a card.
