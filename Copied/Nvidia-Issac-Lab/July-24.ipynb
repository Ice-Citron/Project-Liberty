1. The Philosophy: Escaping Python
   In standard Python, a list is just a collection of pointers--each pointing
   to a separate object scattered somewhere in memory. Looping through a 
   million numbers.... check types and chase pts a mill times... slowl...
   NumPy exists to fix this... takes data and packs into a single contiguous
   block of raw C memory. When add two NumPy arrays, you aren' running a Python
   loop; you are issing a single command to a highly optmised C bacend...
   sweep memory at HW speeds. ... entire job in NumPy is to organise your data 
   so you can apply operations to the whole block at once, completely avoiding
   Python `for` loops. This is called vectorisation.

2. THE ILLUSION: MEMORY and STRIDES
   Because an array is just one flat, linear slab of memory under the hood,
   "multidimensionality" is actually an illusion created by metadata. A 
   NumPy array is really just flat data block wrapped in a small header that
   defines its shape ... and its strides (how many bytes to step forward to find
   the next item in a row or column). This is a massive mental unlock... it
   means ops like `.reshape(), .ravel(), .T()...` are instantaneous and cost
   zero memory. NumPy doesn't move a single byte of the actual data; it just
   hands you a new set of "reading glasses" (new shape and strides)... to look
   at the exact same fla...
