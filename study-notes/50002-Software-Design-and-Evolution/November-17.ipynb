-- A record is a shortway to write a class that only holds data, such as 
   `record WeatherForecast(String summary, int temperature) {}`. Java
   reads the names in the brackets and writes private final fields plus a 
   constructor that sets them all for you ... Each field also gets an
   accessor... with the same name as ... and no `get` in front ...
   Java also writes `equals()` and `hashCode()` and `toString()` from every 
   field so two records with the same values count as equal ... Records are
   final and their fields cannot change, but you can add a compact constructor
   to check the values when one is created ...
