   Your understanding of `pixi run` is correct. It prepares the project 
   environment, including paths, Python, ROS variables, and the activation 
   script. It then runs the rest of the line as one normal command inside that
   environment. This is similar to using `conda activate` and then running a
   command. However, `ros2` is not a Python file argument. It is its own
   executable program and CLI tool that comes with ROS 2. So the command does
   not use ... 
      `pixi run` provides the environment
      `ros2` program runs inside it
      The `run` after `ros2` is a subcommand, like `commit` is sa subcommand of 
      git. It  finds an installed program in a named ROS package and starts it,
      so you do not need to know the program's file path. Everything after 
      `--ros-args` goes to the started program, not to `ros2`. Each #
      `-p name:=value` sets a ROS parameter on the node. Hre, these aprameters 
      set `use_sim_time` and choose which policy class to load.

   The repeated `aic_model aic_model` is not an error. The format is
   `ros2 run <package> <executable>`. 
   The first word is the package name. The second word is the name of an 
   executable in that package. A package can contain many executables. For
   example, the `aic_bringup` package contains `home_robot.py` and 
   `test_impedance.py`. That is why both worlds are needed. Here, the developers
   gave the package and its main executable the same name. This is common in ROS
   , so `aic_model` appears twice. The full command means: inside the pixi
   environment, use the ROS tool to start the `aic_model` program from the 
   `aic_model` package, use simulation time, and load `aic_example_policies.ros.GentleGiant`

   `Feedforward wrench or joint torque` is the anticipatory, open-loop effort
   calculated directly from  the robot's dynamic model and intended task profile
   before any error is measured. While a wrench descrives the 6D generalised 
   spatial force and torque vector ... at a specific frame (like the 
   end-effector) and joint torque refers to the generalised actuator efforts, 
   the feedforward term provides 

1. high level policy
2. trajectory interpolator / motion planner                 creates waypoint for each 1ms tick
3. low-level controller (computers torque && imperance)     consider physics and error correction, outputs desired torque for   
                                                            motors at each ms 
        THIS IS WHERE `feedforward torque` and `feedback` lives.
4. motor driver                                             converts desired torque to winding current comamnds based on motor 
                                                            properties
5. Physical motor movement
