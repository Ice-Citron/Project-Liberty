   ... PAIRED first, mechanically. Standard domain randomisation -- what your
   M2 collection does -- samples scene configurations uniformly from the 
   declared bounds: task board pose anywhere in the range, connectors anywhere
   on their rails, no intelligence about which configured matter. PAIRED
   ... ("unsupervised environment design") replaces the uniform sampler with a
   third agent whose job is to choose the configuration. It's trained to 
   maximise REGRET: the gap between what a reference agent achieves on that
   config and what the training policy achieves. That objective is clever for
   one specific reason -- it can't win by proposing impossible scenarios (if 
   nobody can solve it, regret is zero) -- so it's forced to find configs
   that are solvable but failed: exactly the frontier where the policy is 
   weakes. The policy trains on those, improves, the adversar finds the new 
   frontier, and you get an automatic curriculum that walks the failure 
   boundary. Successors like PLR and ACCEL do the same thing more cheaply by
   replaing and mutating levels with high learning value instead of training
   a full adversary network.

   A "config the policy fails on" is a coverage hole. And your Intrinsic loss
   was this phenomenon: you trained and tested on near-rack poses, the
   eval landed far-rack, and you lost to a region of parameter space nobody
   had searched. An adversary pointed at your DR bounds would have found that
   hole in an afternoon, because far-rack was solvable (the winner solved it)
   and you failed it -- maximal regret, ... as the adversary network:
   propse scenario, run policy, harvest failure, get paid per hole. 

   The duality, then, stated plainly: training and certification run the
   identical search loop -- propose configuration -> execute policy ->
   detect failure -- and differ only in what consumes the output ...
