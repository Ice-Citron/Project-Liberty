   The `tee` command reads standard input and writes it to both standard output
   (terminal screen) and one or more files at the same time.

   It gets its name from plumbing, where a T-shaped pipe splitter splits a 
   single stream of water into two different directions. In the terminal, it 
   allows you to view the live output of a command while continuously saving
   that output to a log file.
