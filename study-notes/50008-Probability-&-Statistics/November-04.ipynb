   The symbol $h$ is just standard calculus shorthand for a tiny horizontal
   step size ($\delta x or dx$), representing the width of a skinny slice along
   the x-axis. Since the exact probability at a single point is zero 
   (`P(X = x) = 0`), the only way to get a real, non-zero probability around $x$
   is to open up a tiny window from $x$ to $x + h$. Over that microscopic width
   $h$, the curve $f_X(x)$ barely has room to change ... so the slice under the 
   curve looks almost exactly like a thin vertical rectangle of height ...
   $f_X(x)$ and base width $h$. Because probability equals area, the area of 
   that skinny rectangle: `height * width = f_X(x) * h`, gives you a quick,
   accurate approximation of the integral ...
