-- WHY IS A COIN TOSS SEQUENCE == A STOCHASTIC PROCESS

   When you repeat a coin toss, each flip produces a random outcome that cannot
   be predictd with certainty ... In probability theory, mathematicians 
   represent each individual flip as a discrete RV taking a value of zero or one
   ... By ordering these flips sequentially, the index of each toss acts as a
   discrete measurement of time ,,, 

   `A STOCHASTIC PROCESS IS SIMPLY A COLLECTION OF RV INDEXED BY TIME,...`
   which matches this setup .. Even though each coin flip is independent of
   the rprevious one, the entire unfolding sequence evolves randomly
   step by step ... 


-- i.i.d == INDEPENDENT AND IDENTICALLY DISTRIBUTED
   ... meaning every coin toss in the sequence behaves like a fresh, identical
   random event with no memory of past outcomes .. 
   
   INDEPENDENT means the result 
   of any step `X_n`has zeo effect on outcome of any later step ...

   IDENTICALLY DISTRIBUTED means every random varaibles `X_0, X_1, X_2, ...`
   follows the exact same probability rrule at every time step, such as 
   $P(X_n = 1) = 1/2$ throughout the entire sequence. Taken together ...
   i.i.d describes a MEMORYLESS PROCESS where time steps do not
   depend oon one another, which contrasts with a `Discrete-Time Markov Chain`
   `(DTMC)` ... where the current state directly shapes the probabilities
   of the next step ...
