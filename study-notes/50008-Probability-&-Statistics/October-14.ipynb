   in maths, saying a collection is "closed under" an operation means that if
   you take items from that collection and apply the operation, the result
   cannot escape the collection... it is guaranteed to still be inside... adding
   two integers... the result is always anothe integer... so integers are closed
   under addition... the large U symbol is just the shorthand... 

   ... the word countable restricts how many sets you are allowed to combine
   at once... it means you can list the sets out in a discrete sequence... the
   first set... as long as you can map the sets to the natural numbers... they 
   are countable... This is a mathematically distinct from uncountable
   infinities... which represent a continuous, un-listable smear of values... 

   Probability theory requires this specific rule because we often need to 
   calculate the chance of a sequence of events occuring over time, such as

   that $\mathcal{F}$ is closed under countable unions, the math guarantees that
