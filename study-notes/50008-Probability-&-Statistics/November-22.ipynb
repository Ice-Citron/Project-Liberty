# Statistical Modelling

   Statistics assumes the availability of a `SAMPLE` of observations
   - A sample is a subset of a POPULATION of interest for the study
   - e.g., a voting poll uses a random sample of the voters ... 

   We focus on `PARAMETRIC STATISTICS`, where the population's data follow some
   distribution ... 

   The goal is then to estiamte the model parameters ... from the sample of 
   observed data.
   - Samples will be affected by uncertainty due to limited sample size and
     due to random selection.

-- i.i.d means every draw comes from the same distribution and no draw 
   affects any othe. 

    Take a bag with 3 red and 2 blue balls: if you keep each ball out, the second draw depends on the first and you can never take more than 5 samples. If you put each ball back then every draw has P(red) = 3/5 again, so the draws are i.i.d. and n can be as large as you like. In real life this fits repeated measurements like server response times or coin flips, where each new measurement is a fresh draw that doesn't use up the source.

-- ... in parametric statistics, `parametric` refers to the parameters of the 
   probability distribution assumed to generate the data ... 

   A parameter is a fixed numerical value (or finite set of values) that defines
   a distribution's shape, center, or spread ...
