   `knapsack wvs c` finds the best total value you can fit into a bag of
   capacity `c`... Each item is a `(name, weight, value)` triple and you can
   take the same item again and again ... For capcity `c` the function tries
   every item that still fits... For each one it adds 

```hs
knapsack :: forall name weight value. (Ord weight, Num weight, Ord value, Num value) => [(name, weight, value)] -> weight -> value
knapsack wvs c = maximumElse 0 [ v + knapsack wvs (c - w) | (_,w,v) <- wvs , w <= c ]
```





```hs
knapsack wvs 3
= maximumElse 0 [1000 + knapsack wvs 0, 500 + knapsack wvs 2]

  knapsack wvs 0
  = maximumElse 0 []                       -- nothing fits
  = 0

  knapsack wvs 2
  = maximumElse 0 [500 + knapsack wvs 1]   -- item 1 does not fit
  = maximumElse 0 [500 + maximumElse 0 [500 + knapsack wvs 0]]
  = maximumElse 0 [500 + maximumElse 0 [500 + 0]]
  = maximumElse 0 [500 + 500]
  = 1000

= maximumElse 0 [1000 + 0, 500 + 1000]
= maximumElse 0 [1000, 1500]
= 1500
```
