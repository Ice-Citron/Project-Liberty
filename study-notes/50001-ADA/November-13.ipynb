Are ships limited by how many turns they have?

No. Ships have no fuel or turn limit. A turn is one tick of game time. Turns on a Wormhole tells you how many ticks it takes to travel through it. Ships you send out form a Fleet. The fleet records how many turns remain before it arrives. Each game turn reduces that number by one. When it reaches zero, the fleet arrives and joins or attacks the planet. So a 20-turn wormhole holds the fleet for 20 ticks, and a 10-turn wormhole holds it for 10 ticks. CW1 never runs this clock. It uses Turns only as the edge weight in Dijkstra, so the shortest path is the fastest one.

Does Ships mean the number of ships I have?

It means the number of ships on one planet, not the total number across the map. The spec calls this the garrison. On a neutral planet, these ships are its defence. You need more attacking ships than defending ships to capture it. That is why knapsack uses Ships as the weight.

What is Growth?

Growth is the number of new ships a planet makes each turn. Only a planet owned by a player makes ships, so a neutral planet’s growth is unused. Think of it as income per turn. If you capture a planet with growth 50, you get 50 extra ships every turn after that. So growth is the planet’s long-term value, and knapsack uses it as the value to maximise.
