```haskell
import qualified Data.Map as M

ages :: M.Map String Int
ages = M.fromList [("ana", 20), ("bo", 31)]

M.lookup "ana" ages   -- Just 20
M.lookup "cy"  ages   -- Nothing
ages M.! "bo"         -- 31        (crashes if the key is missing)
M.keys ages           -- ["ana","bo"]
M.elems ages          -- [20,31]
M.toList ages         -- [("ana",20),("bo",31)]
```
