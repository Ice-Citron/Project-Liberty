THE SYSTEM CALL DISPATCHER & File Descriptor 1
   In PintOS... the SYSTEM CALL DISPATCHER is the central traffic cop for 
   user-to-kernel communication.
   - THE DISPATCHER: When a user program calls a function like `write()` or 
     `exit()`, it triggers a software interrupt (`intr 0x30`). This pauses
     the user program and drops the CPU into kernel mode, landing directly 
     inside `syscall_handler`.... The dispatcher reads the system call number
     from the stack, figures out which operation the user wants, and forwards
     the arguments to the correct kernel function.
   - FILE DESCRIPTOR 1 (`fd 1`): This is `stdout` (STANDARD OUTPUT). In this
     stage of PintOS, when a user program calls `write(1, buffer, size)` it
     expects the text to be printed directly to the console/terminal screen
     using the kernel's `putbuf()` function.
