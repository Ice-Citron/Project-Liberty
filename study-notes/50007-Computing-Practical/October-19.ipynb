   The command `git merge --ff-only origin/master`, it tells Git t bring your
   current local branch up to date with the remote branch (`origin/master`), but
   ONLY IF IT CAN DO A "FAST-FORWARD" MERGE. 

   ...
   - FAST-FORWARD ONLY: This means your local branch has no unique changes or 
     new commits of its own. It is just lagging behind the server. Git can 
     simply "slide" your branch pointer forward to catch up.
   - NO MERGE COMMITS: If someone else has made changes on the server and you
     have also made your own local commits, a fast-foward is impossible. Instead
     of forcing a messy merge or creating a new "merge commmit," this command
     will safely stop and say: `fatal: Not possible to fast-forward, aborting`
