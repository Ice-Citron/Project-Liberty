   MULTIPLE CONATIONS happen when several differnt high-priority threads are
   all waiting on locks held by a signle, low-priority bottleneck thread. For
   example, if a priority-40 thread and a priority-60 thread both want llocks
   that Thread X currently holds, Thread X receives "multiple" donations
   and runs at priority 60 (the maximum of all its donors) until it releases
   the locks. 

   NESTED DONATIONS, on the other hand, forms a chain. If a priority-60 thread
   waits on a lock held by Thread Y, but Thread Y is current stuck on a lock 
   held by Thread Z, the priority must "nest" or propagate down the chain so 
   that Thread Z gets boosted to 60.

---

   ... in the real world, this is called PRIORITY INHERITANCE (or a stricter
   variant called Priority Ceiling Protocol), which mathematically guarantees
   that a higher-piority task will never be indefinitely delayed by mid-priority
   background noise.




   PREEMPTION means forcibly interrupting a currently running thread to give the
   CPU to another thread. Instead of waiting for the running thread to finish or
   voluntarily give up control (yield), the OS steps in, pauses it, saves its
   state, and schedules a higher-priority thread to run. 



   ... when a thread calls `cond_wait`, the PintOS creates a temporary
   `semaphore_elem` on the thread's stack. This struct holds a sempahore 
   initialized to 0. The thread adds this `semaphore_elem` to the Condition
   Variable's list of waiters, and then immediately calls `sema_down()`...
   on that semaphore to put itself to sleep...

   when another thread calls `cond_signal`.... it needs to wake up the highest
   priority THREAD waiting on the condition variable. To do this... it needs
   to compare the priorities of the different `semaphore_elem`s in the list...
