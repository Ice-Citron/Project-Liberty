   `switch_threads()` saves the current registers and `esp` into 
   `cur->stack`, loads `esp` from the `next->stack`, and returns in the next
   thread. `schedule()` calls it; `thread_schedule_tail()` then marks the new
   thread as THREAD_RUNNING. 

---

   Back in, front out gives round-robin. `thread_unblock()` does not pre-empt.
   `schedule()` asserts interrupts are off. The idle thread is never on 
   `ready_list`; it runs only when the list is empty.

---

   PRIu64 comes from `<inttypes.h>` and expands to the right format letters; 
   adjacent string literals join into one format string. PRId64 is for the
   signed `int64_t`.

---

   A lock is a semaphore that starts at 1, plus a holder. A semaphore has no
   owner: any thread can up it. `lock_release` asserts 
   `lock_held_by_current_thread`. PintOS locks are not recursive.

---

   This is a race condition: shared data, at least oone writier, non-atomic
   access, and a timing-dependent result (time of check to time to use). 

   Fix: hold one lock across the check and the use in every thread that touches
   `x`, or turn interrupts off if a handler shares it.
