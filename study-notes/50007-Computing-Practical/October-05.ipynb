   `switch_threads()` saves the current registers and `esp` into 
   `cur->stack`, loads `esp` from the `next->stack`, and returns in the next
   thread. `schedule()` calls it; `thread_schedule_tail()` then marks the new
   thread as THREAD_RUNNING.

   Back in, front out gives round-robin. `thread_unblock()` does not pre-empt.
   `schedule()` asserts interrupts are off. The idle thread is never on 
   `ready_list`; it runs only when the list is empty.
